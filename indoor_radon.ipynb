# Average concentration across various towns

### ZUMBRUGU

```python
bars = plt.bar(SUMBRUGU["sample ID"], SUMBRUGU["Radon Concentration"])

plt.xlabel("Houses at Sumbrugu community")
plt.ylabel(r"Average Conc. $Bq/m^3$")
plt.title("Average Radon Conc. in Houses at Sumbrugu")
plt.xticks(rotation=45)

# Add values on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,  
        height,                           
        f"{height:.2f}",                  
        ha="center", va="bottom"          
    )

plt.show()
```

### Average Radon Concentration of Communities

```python
averages=[SUMBRUGU["Radon Concentration"].mean(),SHERIGU["Radon Concentration"].mean(),YOROGO["Radon Concentration"].mean(),YARIGABISI["Radon Concentration"].mean(),ZUARUNGU["Radon Concentration"].mean(),GAMBIGBO["Radon Concentration"].mean()]
cities=["SUMBRUGU","SHERIGU","YOROGO","YARIGABISI","ZUARUNGU","GAMBIGBO"]
mean={"Communities": cities, "Average conc.":averages}
```

```python
radon=pd.DataFrame(mean)
```

```python
bars=plt.bar(radon["Communities"],radon["Average conc."])
plt.xlabel("Study Area")
plt.ylabel(r"Average concentration $Bq/m^3$")
plt.title("Comaprison of Average radon concentation in communities")
plt.xticks(rotation=45)
for bar in bars:
    height=bar.get_height()
    plt.text(
        bar.get_x()+bar.get_width()/2,height, 
        f"{height:.2f}",
        ha="center",va="bottom"

    )
plt.show()
```
